```python
from google import genai
from google.genai import types
from google.colab import userdata

client = genai.Client(
    api_key=userdata.get("GEMINI_API_KEY")
)

MODEL = "gemini-3.1-flash-lite"
```

**FAQ information**

```python
knowledge_base = """
COMPANY: FoodExpress
Food Delivery Customer Support FAQs

Q: How do I place an order?
A: Open the app, select a restaurant, choose your food items,
add them to your cart, select your delivery address and payment
method, and click Place Order.

Q: How can I track my order?
A: Open My Orders and select your current order to check the
order status and delivery information.

Q: Can I cancel my order?
A: You can cancel your order if the restaurant has not started
preparing it.

Q: My order is delayed. What should I do?
A: Check the estimated delivery time in My Orders. If the order
is significantly delayed, contact customer support through
the Help section.

Q: I received the wrong food. What should I do?
A: Open My Orders, select the order and go to Help. Report the
incorrect item.

Q: How can I request a refund?
A: Open the relevant order, go to Help and submit a refund
request if the order qualifies.

Q: How do I contact customer support?
A: Open the app and go to the Help or Support section.

Q: How do I reset my password?
A: On the login screen, select Forgot Password and follow the
instructions to reset your password.
"""
```

**chatbot instuctions**

```python
system_instruction = f"""

You are a friendly customer support assistant for FoodExpress.

Use the FAQ information given below to answer customer questions.

Answer only using the information provided.

Do not make up information.

If the answer is not available in the FAQ,
say that you don't have enough information.

Keep your answers simple and short.

FAQ INFORMATION:

{knowledge_base}

"""
```

creating chatbot

```python
chat = client.chats.create(
    model=MODEL,
    config=types.GenerateContentConfig(
        system_instruction=system_instruction,
        temperature=0.2,
        max_output_tokens=200,
        thinking_config=types.ThinkingConfig(
            thinking_level="low"
        )
    ),
    history=[]
)

print("Hi! I am FoodExpress Support.")
print("Type 'bye' or 'quit' to exit.\n")

while True:

    user_input = input("Ask your Question: ")

    if user_input.lower() in ["bye", "quit"]:
        print("Thanks for contacting FoodExpress!")
        break

    reply = chat.send_message(user_input)

    print("FoodExpress Support:", reply.text)
    print()
```
